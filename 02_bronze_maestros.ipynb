```python
CATALOGO = "workspace"

productos = [
    # producto_id,   nombre,             linea,        cultivo,  vida_util_dias, sensible_humedad
    ("NITRO-XP",     "NitroXtent XP",     "NitroXtent", "Cana",   180, False),
    ("NITRO-S-XP",   "NitroXtent+S XP",   "NitroXtent", "Cana",   120, True),
    ("FERTI-CAFE",   "Ferticrop Cafe",    "Ferticrop",  "Cafe",    90, True),
    ("FERTI-BANANO", "Ferticrop Banano",  "Ferticrop",  "Banano", 120, True),
    ("FERTI-MAIZ",   "Ferticrop Maiz",    "Ferticrop",  "Maiz",   180, False),
    ("FERTI-PAPA",   "Ferticrop Papa",    "Ferticrop",  "Papa",   150, False),
    ("FERTI-PASTOS", "Ferticrop Pastos",  "Ferticrop",  "Pastos", 180, False),
    ("FERTI-ARROZ",  "Ferticrop Arroz",   "Ferticrop",  "Arroz",   90, True),
]

cols = ["producto_id","nombre","linea","cultivo_objetivo","vida_util_dias","sensible_humedad"]
df_prod = spark.createDataFrame(productos, cols)

df_prod.write.format("delta").mode("overwrite").saveAsTable(f"{CATALOGO}.bronze.productos_raw")
display(spark.table(f"{CATALOGO}.bronze.productos_raw"))
```

```python
import random
random.seed(7)

tipos  = ["Ingenio","Cooperativa Cafe","Cooperativa Banano","Cooperativa Arroz","Granja","Distribuidor","Finca"]
deptos = ["Valle","Tolima","Antioquia","Quindio","Narino"]

clientes = []
for i in range(1, 41):                      
    cid = f"C{str(i).zfill(3)}"
    clientes.append((cid, f"Cliente {cid}", random.choice(tipos), random.choice(deptos)))

df_cli = spark.createDataFrame(clientes, ["cliente_id","nombre","tipo","departamento"])
df_cli.write.format("delta").mode("overwrite").saveAsTable(f"{CATALOGO}.bronze.clientes_raw")
display(spark.table(f"{CATALOGO}.bronze.clientes_raw"))
```
